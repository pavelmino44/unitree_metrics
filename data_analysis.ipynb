```python
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator
import os
import matplotlib.pyplot as plt
```

## Путь к логам

```python
run_1 = "logs/run_1"
```

```python
run_2 = "logs/run_2"
```

## Находим TensorBoard и загружаем лог

```python
event_file = next(
    os.path.join(run_1, f)
    for f in os.listdir(run_1)
    if f.startswith("events.out")
)

ea_1 = EventAccumulator(event_file)
ea_1.Reload()
```

```python
event_file = next(
    os.path.join(run_2, f)
    for f in os.listdir(run_2)
    if f.startswith("events.out")
)

ea_2 = EventAccumulator(event_file)
ea_2.Reload()
```

## Что есть в TensorBoard

```python
ea_1.Tags()
```

```python
scalar_tags = ea_1.Tags()["scalars"]

print(f"Всего scalar-метрик: {len(scalar_tags)}\n")

for i, tag in enumerate(scalar_tags):
    print(f"{i:2d}: {tag}")
```

## Поиск нужных метрник

```python
for tag in scalar_tags:
    if "reward" in tag.lower():
        print(tag)

print("\n--------------------\n")

for tag in scalar_tags:
    if "episode" in tag.lower():
        print(tag)

print("\n--------------------\n")

for tag in scalar_tags:
    if "length" in tag.lower():
        print(tag)
```

## Смотрим шаги обучения

### Первый этап

```python
reward = ea_1.Scalars("Train/mean_reward")

print("Количество точек:", len(reward))

print("\nПервые 10:")
for s in reward[:10]:
    print(s.step, s.value)

print("\nПоследние 10:")
for s in reward[-10:]:
    print(s.step, s.value)
```

### Второй этап

```python
reward = ea_2.Scalars("Train/mean_reward")

print("Количество точек:", len(reward))

print("\nПервые 10:")
for s in reward[:10]:
    print(s.step, s.value)

print("\nПоследние 10:")
for s in reward[-10:]:
    print(s.step, s.value)
```
